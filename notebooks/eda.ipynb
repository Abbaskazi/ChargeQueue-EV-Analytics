# ChargeQueue Phase 3: Exploratory Data Analysis

This notebook documents the reproducible EDA workflow for the processed EV charging dataset. It uses `ports_occupied` as the primary demand variable and keeps zero-denominator operational-utilization rows out of utilization statistics. It does not train forecasting models or split the data.

## Scope and inputs

The notebook reads `data/processed/ev_charging_processed.csv` and writes analysis artifacts under `data/reports/eda/`. The raw and processed datasets are never modified.

```python
from pathlib import Path
import sys
import json
import pandas as pd

sys.path.insert(0, '..')
from backend.app.services.eda import run_eda

processed_path = Path('../data/processed/ev_charging_processed.csv')
output_dir = Path('../data/reports/eda')
result = run_eda(processed_path, output_dir)
result['overview']
```

## Dataset overview

```python
overview = json.loads((output_dir / 'dataset_overview.json').read_text())
pd.Series(overview)
```

## Temporal analysis

```python
hourly = pd.read_csv(output_dir / 'hourly_demand.csv')
weekday = pd.read_csv(output_dir / 'weekday_demand.csv')
monthly = pd.read_csv(output_dir / 'monthly_demand.csv')
hourly.head(), weekday, monthly
```

## Station-wise analysis

```python
station_summary = pd.read_csv(output_dir / 'station_demand_summary.csv')
top_stations = pd.read_csv(output_dir / 'top_10_stations.csv')
top_utilization = pd.read_csv(output_dir / 'top_10_utilization.csv')
top_stations.head(), top_utilization.head()
```

## Distribution and outliers

```python
frequency = pd.read_csv(output_dir / 'occupancy_frequency.csv')
outliers = json.loads((output_dir / 'outlier_summary.json').read_text())
frequency, pd.Series(outliers)
```

## Seasonality and forecasting suitability

The generated heatmap and weekday/weekend comparison are descriptive evidence of recurring patterns. The forecasting suitability report records the regularity, variation, limitations, and any available illustrative stationarity checks. No forecasting model is trained in this notebook.
